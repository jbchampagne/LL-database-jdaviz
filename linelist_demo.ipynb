```python
# read in individual files
line_list = {}
all_linelists = glob('*.csv')
for a in all_linelists:
    this_name = a.split('.csv')[0]
    print(this_name)
    this_schema = schema['extra_info_columns_by_file'][a]
    this_list = ingest.read_one_file(a, this_schema)
    line_list[this_name] = this_list
```
